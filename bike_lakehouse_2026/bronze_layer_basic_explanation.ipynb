# Step 1: Reading from CSV File

```python
df = (
    spark.read.option('header', 'true')
    .option('inferSchema', 'true')
    .csv('/Volumes/baraa_dev_project/bronze/source_system/source_crm/cust_info.csv')
)
df.display()
```
