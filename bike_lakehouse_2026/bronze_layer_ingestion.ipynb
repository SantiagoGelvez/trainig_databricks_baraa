```python
from bronze_ingest_config import SOURCE_CONFIG

for source, raw_data in SOURCE_CONFIG.items():
    print(f"loading {source} data...")
    for file in raw_data:
        print(f"Ingesting {file['path']} -> bronze.{file['table']}")
        df = (
            spark.read
                .option('header', 'true')
                .option('inferSchema', 'true')
                .csv(file['path'])
        )

        (
            df.write
                .mode('overwrite')
                .format('delta')
                .saveAsTable(f'baraa_dev_project.bronze.{file["table"]}')
        )
```
